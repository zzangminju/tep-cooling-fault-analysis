```python
import numpy as np
import pandas as pd


def calculate_kpis(result, fault, seed):

    time = np.asarray(result.time)

    # XMEAS(9) = Reactor Temperature
    temp = result.measurements[:, 8]

    # XMV(10) = Reactor Cooling Water control
    xmv10 = result.manipulated_vars[:, 9]


    # -------------------------
    # 1. 정상 구간: 0 ~ 8 h
    # -------------------------
    normal_mask = time < 8

    normal_temp = temp[normal_mask]
    normal_xmv10 = xmv10[normal_mask]

    normal_temp_mean = normal_temp.mean()
    normal_temp_std = normal_temp.std(ddof=1)

    lower = normal_temp_mean - 2 * normal_temp_std
    upper = normal_temp_mean + 2 * normal_temp_std

    normal_xmv10_mean = normal_xmv10.mean()


    # -------------------------
    # 2. Fault 이후 첫 2시간
    #    8 <= time < 10
    # -------------------------
    fault_mask = (time >= 8) & (time < 10)

    fault_time = time[fault_mask]
    fault_temp = temp[fault_mask]
    fault_xmv10 = xmv10[fault_mask]


    # 데이터 간격: 3분 = 0.05시간
    dt_h = 3 / 60


    # -------------------------
    # 3. Recovery Time
    # -------------------------
    # Fault 발생 직후가 아니라,
    # 8시간 이후 처음 정상범위로 들어오는 시점
    after_fault_mask = time > 8

    after_time = time[after_fault_mask]
    after_temp = temp[after_fault_mask]

    in_range_after = (
        (after_temp >= lower) &
        (after_temp <= upper)
    )

    if in_range_after.any():
        first_recovery_time = after_time[
            np.where(in_range_after)[0][0]
        ]

        recovery_min = (first_recovery_time - 8) * 60
    else:
        recovery_min = np.nan


    # -------------------------
    # 4. Settling Time
    # -------------------------
    # 30분 = 연속 10개 sample
    settling_min = np.nan

    required_samples = 10

    for i in range(
        len(in_range_after) - required_samples + 1
    ):

        if in_range_after[
            i:i + required_samples
        ].all():

            settling_min = (
                after_time[i] - 8
            ) * 60

            break


    # -------------------------
    # 5. In-range Ratio
    # -------------------------
    fault_in_range = (
        (fault_temp >= lower) &
        (fault_temp <= upper)
    )

    in_range_ratio = (
        fault_in_range.mean() * 100
    )


    # -------------------------
    # 6. IAE / ISE
    # -------------------------
    error = fault_temp - normal_temp_mean

    iae = (
        np.abs(error) * dt_h
    ).sum()

    ise = (
        error ** 2 * dt_h
    ).sum()


    # -------------------------
    # 7. Control Burden
    # -------------------------
    control_burden = (
        np.abs(
            fault_xmv10 - normal_xmv10_mean
        ) * dt_h
    ).sum()


    return {
        "Fault": fault,
        "Seed": seed,
        "Recovery_min": recovery_min,
        "Settling_min": settling_min,
        "InRange_%": in_range_ratio,
        "IAE": iae,
        "ISE": ise,
        "Control_Burden": control_burden
    }


# ---------------------------------
# Fault 4 / Fault 14, seed 1~3 계산
# ---------------------------------

kpi_rows = []

for fault in [4, 14]:

    for seed in [1, 2, 3]:

        row = calculate_kpis(
            results[fault][seed],
            fault,
            seed
        )

        kpi_rows.append(row)


kpi_df = pd.DataFrame(kpi_rows)

kpi_df
```

```python
def calculate_kpis(result, fault, seed):

    # 부동소수점 오차 제거
    time = np.round(
        np.asarray(result.time),
        8
    )

    # XMEAS(9): Reactor Temperature
    temp = result.measurements[:, 8]

    # XMV(10): Reactor Cooling Water control
    xmv10 = result.manipulated_vars[:, 9]

    fault_onset = 8.0
    evaluation_end = 10.0

    dt_h = 3 / 60


    # -------------------------
    # 1. 정상 구간: 0 <= t < 8
    # -------------------------
    normal_mask = time < fault_onset

    normal_temp = temp[normal_mask]
    normal_xmv10 = xmv10[normal_mask]

    normal_temp_mean = normal_temp.mean()
    normal_temp_std = normal_temp.std(ddof=1)

    lower = (
        normal_temp_mean
        - 2 * normal_temp_std
    )

    upper = (
        normal_temp_mean
        + 2 * normal_temp_std
    )

    normal_xmv10_mean = (
        normal_xmv10.mean()
    )


    # -------------------------
    # 2. Fault 첫 2시간
    #    8 <= t < 10
    # -------------------------
    fault_mask = (
        (time >= fault_onset)
        & (time < evaluation_end)
    )

    fault_temp = temp[fault_mask]
    fault_xmv10 = xmv10[fault_mask]


    # -------------------------
    # 3. Recovery Time
    # Fault 발생 다음 sample부터 확인
    # -------------------------
    after_fault_mask = (
        time > fault_onset
    )

    after_time = time[after_fault_mask]
    after_temp = temp[after_fault_mask]

    in_range_after = (
        (after_temp >= lower)
        & (after_temp <= upper)
    )

    if in_range_after.any():

        first_idx = np.where(
            in_range_after
        )[0][0]

        recovery_min = (
            after_time[first_idx]
            - fault_onset
        ) * 60

    else:
        recovery_min = np.nan


    # -------------------------
    # 4. Settling Time
    # 30분 연속 = 10 samples
    # -------------------------
    settling_min = np.nan

    required_samples = 10

    for i in range(
        len(in_range_after)
        - required_samples
        + 1
    ):

        if in_range_after[
            i:i + required_samples
        ].all():

            settling_min = (
                after_time[i]
                - fault_onset
            ) * 60

            break


    # -------------------------
    # 5. In-range Ratio
    # -------------------------
    fault_in_range = (
        (fault_temp >= lower)
        & (fault_temp <= upper)
    )

    in_range_ratio = (
        fault_in_range.mean()
        * 100
    )


    # -------------------------
    # 6. IAE / ISE
    # -------------------------
    error = (
        fault_temp
        - normal_temp_mean
    )

    iae = (
        np.abs(error) * dt_h
    ).sum()

    ise = (
        error**2 * dt_h
    ).sum()


    # -------------------------
    # 7. Control Burden
    # -------------------------
    control_burden = (
        np.abs(
            fault_xmv10
            - normal_xmv10_mean
        ) * dt_h
    ).sum()


    return {
        "Fault": fault,
        "Seed": seed,
        "Recovery_min": recovery_min,
        "Settling_min": settling_min,
        "InRange_%": in_range_ratio,
        "IAE": iae,
        "ISE": ise,
        "Control_Burden": control_burden
    }
```

```python
import numpy as np
import pandas as pd


def calculate_kpis_final(result, fault, seed):

    time = np.round(
        np.asarray(result.time),
        8
    )

    # XMEAS(9): Reactor Temperature
    temp = result.measurements[:, 8]

    # XMV(10): Reactor Cooling Water control
    xmv10 = result.manipulated_vars[:, 9]

    fault_onset = 8.0

    # IAE / ISE / Control Burden 평가구간
    evaluation_end = 10.0

    # 시뮬레이션 전체 종료시간
    # Settling은 12 h까지 관찰
    settling_end = 12.0

    dt_h = 3 / 60


    # =====================================
    # 1. 정상 구간
    # =====================================

    normal_mask = (
        (time >= 0)
        & (time < fault_onset)
    )

    normal_temp = temp[normal_mask]
    normal_xmv10 = xmv10[normal_mask]

    normal_temp_mean = normal_temp.mean()
    normal_temp_std = normal_temp.std(ddof=1)

    lower = (
        normal_temp_mean
        - 2 * normal_temp_std
    )

    upper = (
        normal_temp_mean
        + 2 * normal_temp_std
    )

    normal_xmv10_mean = (
        normal_xmv10.mean()
    )


    # =====================================
    # 2. Fault 후 첫 2시간
    # =====================================

    fault_mask = (
        (time >= fault_onset)
        & (time < evaluation_end)
    )

    fault_temp = temp[fault_mask]
    fault_xmv10 = xmv10[fault_mask]


    # =====================================
    # 3. Recovery Time
    # =====================================

    recovery_mask = (
        (time > fault_onset)
        & (time <= settling_end)
    )

    recovery_time = time[recovery_mask]
    recovery_temp = temp[recovery_mask]

    in_range_after = (
        (recovery_temp >= lower)
        & (recovery_temp <= upper)
    )

    if in_range_after.any():

        first_idx = np.where(
            in_range_after
        )[0][0]

        recovery_min = (
            recovery_time[first_idx]
            - fault_onset
        ) * 60

    else:
        recovery_min = np.nan


    # =====================================
    # 4. Settling Time
    # 30분 연속 정상범위
    # =====================================

    settling_min = np.nan

    required_samples = 10

    for i in range(
        len(in_range_after)
        - required_samples
        + 1
    ):

        if in_range_after[
            i:i + required_samples
        ].all():

            settling_min = (
                recovery_time[i]
                - fault_onset
            ) * 60

            break


    settled_4h = int(
        not np.isnan(settling_min)
    )


    # =====================================
    # 5. In-range Ratio
    # =====================================

    fault_in_range = (
        (fault_temp >= lower)
        & (fault_temp <= upper)
    )

    in_range_ratio = (
        fault_in_range.mean()
        * 100
    )


    # =====================================
    # 6. Fault IAE / ISE
    # =====================================

    error = (
        fault_temp
        - normal_temp_mean
    )

    fault_iae = (
        np.abs(error)
        * dt_h
    ).sum()

    fault_ise = (
        error**2
        * dt_h
    ).sum()


    # =====================================
    # 7. Fault Control Burden
    # =====================================

    fault_control_burden = (
        np.abs(
            fault_xmv10
            - normal_xmv10_mean
        )
        * dt_h
    ).sum()


    # =====================================
    # 8. 정상 2시간 구간의 기준값
    # =====================================

    normal_iae_list = []
    normal_ise_list = []
    normal_cb_list = []

    normal_windows = [
        (0, 2),
        (2, 4),
        (4, 6),
        (6, 8)
    ]

    for start, end in normal_windows:

        window_mask = (
            (time >= start)
            & (time < end)
        )

        temp_window = temp[window_mask]
        xmv_window = xmv10[window_mask]

        window_error = (
            temp_window
            - normal_temp_mean
        )

        normal_iae = (
            np.abs(window_error)
            * dt_h
        ).sum()

        normal_ise = (
            window_error**2
            * dt_h
        ).sum()

        normal_cb = (
            np.abs(
                xmv_window
                - normal_xmv10_mean
            )
            * dt_h
        ).sum()

        normal_iae_list.append(
            normal_iae
        )

        normal_ise_list.append(
            normal_ise
        )

        normal_cb_list.append(
            normal_cb
        )


    normal_iae_mean = np.mean(
        normal_iae_list
    )

    normal_ise_mean = np.mean(
        normal_ise_list
    )

    normal_cb_mean = np.mean(
        normal_cb_list
    )


    # =====================================
    # 9. 정상 대비 비율
    # =====================================

    iae_ratio = (
        fault_iae
        / normal_iae_mean
    )

    ise_ratio = (
        fault_ise
        / normal_ise_mean
    )

    control_ratio = (
        fault_control_burden
        / normal_cb_mean
    )


    # =====================================
    # 10. 결과 반환
    # =====================================

    return {

        "Fault": fault,
        "Seed": seed,

        "Recovery_min": recovery_min,
        "Settling_min": settling_min,
        "Settled_4h": settled_4h,

        "InRange_%": in_range_ratio,

        "IAE": fault_iae,
        "ISE": fault_ise,
        "Control_Burden": fault_control_burden,

        "IAE_Ratio": iae_ratio,
        "ISE_Ratio": ise_ratio,
        "Control_Ratio": control_ratio
    }
```

```python
from pathlib import Path

Path("ack_results").mkdir(
    exist_ok=True
)


final_rows = []

final_seeds = range(
    101,
    131
)


for fault in [4, 14]:

    for seed in final_seeds:

        print(
            f"Fault {fault}, "
            f"Seed {seed} 실행 중..."
        )

        sim = TEPSimulator(
            random_seed=seed,
            backend="python"
        )

        sim.initialize()

        result = sim.simulate(
            duration_hours=12.0,
            disturbances={
                fault: (8.0, 1)
            },
            record_interval=180
        )

        row = calculate_kpis_final(
            result,
            fault,
            seed
        )

        row["Shutdown"] = (
            result.shutdown
        )

        final_rows.append(
            row
        )

        print(
            "완료 |",
            "shutdown =",
            result.shutdown
        )


final_kpi_df = pd.DataFrame(
    final_rows
)


# CSV로 저장
final_kpi_df.to_csv(
    "ack_results/kpi_30runs.csv",
    index=False
)


print()
print("전체 완료")
print(
    "총 run 수:",
    len(final_kpi_df)
)

final_kpi_df.head()
```

```python
# Fault별 30-run 요약 통계

summary_rows = []

for fault in [4, 14]:

    group = final_kpi_df[
        final_kpi_df["Fault"] == fault
    ]

    row = {
        "Fault": fault,

        "Recovery Median (min)":
            group["Recovery_min"].median(),

        "Recovery IQR (min)":
            group["Recovery_min"].quantile(0.75)
            - group["Recovery_min"].quantile(0.25),

        "InRange Median (%)":
            group["InRange_%"].median(),

        "InRange IQR (%)":
            group["InRange_%"].quantile(0.75)
            - group["InRange_%"].quantile(0.25),

        "IAE Median":
            group["IAE"].median(),

        "IAE IQR":
            group["IAE"].quantile(0.75)
            - group["IAE"].quantile(0.25),

        "ISE Median":
            group["ISE"].median(),

        "ISE IQR":
            group["ISE"].quantile(0.75)
            - group["ISE"].quantile(0.25),

        "Control Burden Median":
            group["Control_Burden"].median(),

        "Control Burden IQR":
            group["Control_Burden"].quantile(0.75)
            - group["Control_Burden"].quantile(0.25),

        "Settled within 4h (%)":
            group["Settled_4h"].mean() * 100
    }

    summary_rows.append(row)


summary_df = pd.DataFrame(summary_rows)

summary_df.round(4)
```
